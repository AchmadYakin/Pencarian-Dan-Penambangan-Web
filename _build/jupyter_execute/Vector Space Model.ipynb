## Vector Space Model (VSM)

**Vector Space Model** adalah suatu pendekatan matematis yang memetakan dokumen atau query ke dalam ruang vektor berdimensi tinggi, memungkinkan perhitungan kemiripan antara teks-teks tersebut. Model ini digunakan dalam pengolahan teks, pencarian informasi, dan pemrosesan bahasa alami (Natural Language Processing).

### Konsep Term Frequency-Inverse Document Frequency (TF-IDF)

TF-IDF (Term Frequency-Inverse Document Frequency) adalah sebuah metode statistik yang digunakan untuk mengukur pentingnya suatu kata dalam sebuah dokumen dalam kumpulan dokumen yang lebih besar. TF-IDF akan memberikan bobot pada kata-kata yang lebih informatif atau penting dalam sebuah dokumen.

$$
\text{TF-IDF}(t,d) = \text{TF}(t,d) \times \text{IDF}(t)
$$

Di mana:
- \( \text{TF}(t,d) \) adalah frekuensi kemunculan kata \( t \) dalam dokumen \( d \).
- \( \text{IDF}(t) \) adalah bobot dari kata \( t \) dalam seluruh koleksi dokumen.

#### Term Frequency (TF)

Term Frequency (TF) adalah salah satu metode yang digunakan untuk menghitung seberapa sering sebuah kata (term) muncul dalam sebuah dokumen, semakin sering sebuah kata muncul dalam sebuah dokumen, semakin penting kata tersebut dalam konteks dokumen tersebut. TF hanya mempertimbangkan frekuensi kata dalam satu dokumen tanpa membandingkannya dengan dokumen lain.

$$
\text{TF}(t,d) = \frac{f_{t,d}}{N_d}
$$

Di mana:
- \( f_{t,d} \) adalah jumlah kemunculan kata \( t \) dalam dokumen \( d \).
- \( N_d \) adalah jumlah total kata dalam dokumen \( d \).

#### Inverse Document Frequency

Inverse Document Frequency (IDF) adalah ukuran seberapa penting sebuah kata dalam keseluruhan koleksi dokumen. IDF digunakan untuk menurunkan bobot kata-kata umum yang muncul di banyak dokumen dan memberikan bobot lebih tinggi pada kata-kata yang jarang muncul di seluruh koleksi dokumen. semakin jarang sebuah kata muncul di dokumen lain maka semakin penting kata.

$$
\text{IDF}(t) = \log\left(\frac{N}{1 + |d \in D : t \in d|}\right)
$$

Di mana:
- \( N \) adalah total jumlah dokumen dalam koleksi.
- \( |d \in D : t \in d| \) adalah jumlah dokumen yang mengandung kata \( t \).
- Penambahan 1 pada penyebut untuk menghindari pembagian dengan nol.

#### Tujuan Pembobotan Term Dengan TF-IDF

Tujuan dari Pembobotan ini adalah untuk memberikan pembobotan untuk mengetahui seberapa penting suatu kata (term) dalam dokumen dan koleksi dokumen. TF-IDF sebagai berikut :
* Term frequency (TF), yang mengukur seberapa sering suatu term muncul dalam dokumen. Semakin sering term muncul, semakin penting term tersebut untuk dokumen.
* Inverse document frequency (IDF), yang mengukur seberapa umum suatu term muncul dalam koleksi dokumen. Semakin jarang term muncul, semakin informatif term tersebut untuk membedakan dokumen.
